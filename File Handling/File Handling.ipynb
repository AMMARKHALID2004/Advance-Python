```python
#open() opens the file by default in the read mode (r is used for read,w is used for write,r+ is used for read and write,a is used for appending)
f=open('Test.txt')
print(f.name)
f.close()
```

```python
#we should use context manager for doing this 
#for smaller files this method is good
with open("Test.txt",'r') as f:
    f_contents=f.read()
    print(f_contents)


    
```

```python
#readlines() gives a list of the lines of the file
with open('Test.txt') as f:
    f_contents=f.readlines()
    print(f_contents)
```

```python
#readline() reads one line from the file and then move on to the next line
with open("Test.txt") as f:
    f_contents=f.readline()
    print(f_contents,end=' ')
    f_contents=f.readline()
    print(f_contents,end=' ')
    f_contents=f.readline()
    print(f_contents,end=' ')
    
```

```python
#above method can cause memory issues if we tried to read a large file at once and then tried to
#store it in a variable so we can use a loop for reading a large file

with open('Test.txt') as f:
    for line in f:
        print(line)
```

```python
#we can also read as much characters as we want
with open("Test.txt") as f:
    f_contents=f.read(10)
    print(f_contents)
    f_contents=f.read(15)
    print(f_contents)
```

```python
#we can also use a while loop for reading 

size_to_read=10
with open('Test.txt') as f:
    f_contents=f.read(size_to_read)
    while len(f_contents)>0:
        print(f_contents)
        f_contents=f.read(size_to_read)

#At the end the f_contents will have an empty string and it will not satisfy the loop condition
```

```python
#f.tell() tells the current index at which the read pointer is
with open("Test.txt") as f:
    print(f.tell())
```

```python
#seek() sets the tell read index to a specific position

with open('Test.txt') as f:
    f_contents=f.read(10)
    print(f_contents)
    f.seek(0)
    f_contents=f.read(10)
    print(f_contents)
```

```python
#if we open the file in write mode, it will create the file if it does not already exists and if it exists it will overwrite 
#the file
with open('test2.txt','w') as f:
    f.write("Hello World")
    f.seek(0)
    f.write("John Cena")
```

```python
#reading from a file and writing its data to another file
with open('Test.txt','r') as rf:
    with open('test2.txt','w') as wf:
        for line in rf:
            wf.write(line)
```

```python
#for reading from pictures , open the file in the binary mode
with open('Test.txt','rb') as f:
    for line in f:
        print(line)
```
